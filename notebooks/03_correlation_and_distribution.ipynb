# Relationships and distributions

Scatter/bubble/hexbin/contour for two variables; histograms, KDEs, ridgelines, swarms, box and violin plots for one. The seaborn backend adds its statistical estimation (CI bands, KDEs, regression fits).

```python
import warnings
warnings.filterwarnings("ignore", category=UserWarning)
import pandas as pd
import pychartist as pc
import plotly.io as pio
pio.renderers.default = "plotly_mimetype+notebook_connected"   # renders in JupyterLab/VS Code/Colab; plotly.js from CDN
```

```python
pen = pc.datasets.load('penguins'); g07 = pc.datasets.load('gapminder').query('year == 2007')
quakes = pc.datasets.load('earthquakes')
```

```python
c = pc.scatter(pen, 'flipper_length_mm', 'body_mass_g', 'species', trendline='ols', show_stats=True)
print(c.meta['stats'])
c
```

```python
pc.scatter(pen, 'flipper_length_mm', 'body_mass_g', 'species', trendline='lowess', backend='seaborn')
```

```python
pc.bubble(g07, 'gdpPercap', 'lifeExp', 'pop', 'continent', label=None, log_x=True, backend='plotly',
          title='Wealth, health and population (2007)')
```

```python
pc.examples.demo('connected_scatter')
```

## Too many points? Bin them.

```python
pc.hexbin(quakes, 'Longitude', 'Latitude', gridsize=45, title='23k earthquakes, hexagonal bins')
```

```python
pc.contour(pen, 'bill_length_mm', 'bill_depth_mm', points=True, backend='seaborn')
```

```python
pc.heatmap(pen, corr=True, method='spearman', title='Spearman correlation')
```

```python
pc.joint_plot(pen, 'bill_length_mm', 'bill_depth_mm', 'species')
```

```python
pc.pair_plot(pen, ['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm'], 'species')
```

## Distributions

```python
pc.histogram(pen, 'flipper_length_mm', 'species', kde=True, backend='seaborn')
```

```python
pc.density(pen, 'flipper_length_mm', 'species', backend='plotly')
```

```python
pc.examples.demo('ridgeline')
```

```python
pc.beeswarm(pen, 'species', 'body_mass_g', 'sex', title='Every penguin, no overlap')
```

```python
pc.box(pen, 'species', 'body_mass_g', points='all', show_stats=True, backend='plotly')
```

```python
pc.violin(pen, 'species', 'flipper_length_mm', 'sex', split=True, backend='seaborn')
```
